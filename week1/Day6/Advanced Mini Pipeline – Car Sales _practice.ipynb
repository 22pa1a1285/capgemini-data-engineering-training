Phase 2 – Cleaning

```python
from pyspark.sql.functions import col, when, trim

# 1. Handle nulls
customers = customers.fillna({
    "name": "unknown",
    "city": "unknown"
})

cars = cars.fillna({
    "price": 0
})

sales = sales.fillna({
    "quantity": 0
})

dealers = dealers.fillna({
    "name": "unknown",
    "city": "unknown"
})

# 2. Fix negative prices
cars = cars.withColumn(
    "price",
    when(col("price") < 0, None).otherwise(col("price"))
)

# 3. Trim strings
customers = customers.withColumn("name", trim(col("name"))) \
                     .withColumn("city", trim(col("city")))

dealers = dealers.withColumn("name", trim(col("name"))) \
                 .withColumn("city", trim(col("city")))

# 4. Remove invalid foreign keys with aliased columns
sales = sales.join(
    customers.select("customer_id", col("name").alias("customer_name"), col("city").alias("customer_city")),
    "customer_id", "inner"
).join(cars, "car_id", "inner") \
 .join(sales_dealer, "sale_id", "inner") \
 .join(
     dealers.select("dealer_id", col("name").alias("dealer_name"), col("city").alias("dealer_city")),
     "dealer_id", "inner"
 )
```

Phase 3 – Validation

```python
## Invalid customer_id in sales
invalid_customers = sales.join(
    customers,
    on="customer_id",
    how="left_anti"
)
```

```python
## Invalid car_id in sales
invalid_cars = sales.join(
    cars,
    on="car_id",
    how="left_anti"
)
```

```python
## Invalid sale_id in sales_dealer
invalid_sales = dealers.join(
    sales,
    on="sale_id",
    how="left_anti"
)
```

```python
## Invalid dealer_id in sales_dealer
invalid_dealers = sales.join(
    dealers,
    on="dealer_id",
    how="left_anti"
)
```

```python
## Invalid dealer_id in sales_dealer
invalid_dealers = sales_dealer.join(
    dealers,
    on="dealer_id",
    how="left_anti"
)
```

```python
from pyspark.sql.functions import sum, desc, row_number, month, countDistinct
from pyspark.sql.window import Window

# Phase 3 – Validation
# 1. Find invalid FK using left_anti join
invalid_customers = sales.join(customers, on="customer_id", how="left_anti")
invalid_cars = sales.join(cars, on="car_id", how="left_anti")
invalid_dealers = sales_dealer.join(dealers, on="dealer_id", how="left_anti")
invalid_sales_dealer = sales_dealer.join(sales, on="sale_id", how="left_anti")
invalid_dealers_sales_dealer = sales_dealer.join(dealers, on="dealer_id", how="left_anti")

```

```python
# 2. Create validation report
validation_report = spark.createDataFrame([
    ("invalid_customers", invalid_customers.count()),
    ("invalid_cars", invalid_cars.count()),
    ("invalid_dealers", invalid_dealers.count()),
    ("invalid_sales_dealer", invalid_sales_dealer.count()),
    ("invalid_dealers_sales_dealer", invalid_dealers_sales_dealer.count())
], ["validation_type", "invalid_count"])
display(validation_report)
```

```python
# Phase 4 – Transformation
# 1. Customer revenue
customer_revenue = sales.groupBy("customer_id").agg(sum(col("price") * col("quantity")).alias("revenue"))
display(customer_revenue)

# 2. Brand-wise sales
brand_sales = sales.groupBy("brand").agg(sum(col("quantity")).alias("total_sales"))
display(brand_sales)

# 3. City-wise revenue (customer city)
city_revenue = sales.groupBy("customer_city").agg(sum(col("price") * col("quantity")).alias("revenue"))
display(city_revenue)

```

```python
# Phase 5 – Dealer Analytics
# 1. Revenue per dealer
dealer_revenue = sales.groupBy("dealer_id").agg(sum(col("price") * col("quantity")).alias("revenue"))
display(dealer_revenue)

# 2. Top dealers by revenue
top_dealers = dealer_revenue.orderBy(desc("revenue")).limit(5)
display(top_dealers)
```

```python
# 3. Dealer-city contribution
dealer_city_contribution = sales.groupBy("dealer_id", "dealer_city").agg(sum(col("price") * col("quantity")).alias("revenue"))
display(dealer_city_contribution)
```

```python
# Phase 6 – SQL Analysis
sales.createOrReplaceTempView("sales")
customers.createOrReplaceTempView("customers")


```

```python
# 1. Top 3 customers per city
top_customers_city = spark.sql("""
SELECT customer_city, customer_id, revenue
FROM (
  SELECT customer_city, customer_id, SUM(price * quantity) AS revenue,
         ROW_NUMBER() OVER (PARTITION BY customer_city ORDER BY SUM(price * quantity) DESC) as rn
  FROM sales
  GROUP BY customer_city, customer_id
)
WHERE rn <= 3
""")
display(top_customers_city)
```

```python
# 2. Monthly trends
monthly_trends = sales.groupBy(month("sale_date").alias("month")).agg(sum(col("price") * col("quantity")).alias("revenue"))
display(monthly_trends)
```

```python
# 3. Repeat customers
repeat_customers = sales.groupBy("customer_id").agg(countDistinct("sale_date").alias("purchase_count")).filter(col("purchase_count") > 1)
display(repeat_customers)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window

# Load all datasets
base_path = "/Volumes/workspace/default/phase5_practice/archive/"

customers = spark.read.csv(f"{base_path}olist_customers_dataset.csv", header=True, inferSchema=True)
geolocation = spark.read.csv(f"{base_path}olist_geolocation_dataset.csv", header=True, inferSchema=True)
order_items = spark.read.csv(f"{base_path}olist_order_items_dataset.csv", header=True, inferSchema=True)
payments = spark.read.csv(f"{base_path}olist_order_payments_dataset.csv", header=True, inferSchema=True)
reviews = spark.read.csv(f"{base_path}olist_order_reviews_dataset.csv", header=True, inferSchema=True)
orders = spark.read.csv(f"{base_path}olist_orders_dataset.csv", header=True, inferSchema=True)
products = spark.read.csv(f"{base_path}olist_products_dataset.csv", header=True, inferSchema=True)
sellers = spark.read.csv(f"{base_path}olist_sellers_dataset.csv", header=True, inferSchema=True)
category_translation = spark.read.csv(f"{base_path}product_category_name_translation.csv", header=True, inferSchema=True)

print("✓ Data loaded successfully")
```

## Task 1: Top 3 Customers per City


```python
# Step 1: Join payments with orders
payments_orders = payments.join(orders, "order_id")
customer_spend = payments_orders.groupBy("customer_id") \
    .agg(sum("payment_value").alias("total_spend"))
customer_city_spend = customer_spend.join(customers, "customer_id") \
    .select("customer_city", "customer_id", "total_spend")
window_city = Window.partitionBy("customer_city") \
    .orderBy(col("total_spend").desc())
customer_city_ranked = customer_city_spend \
    .withColumn("rank", rank().over(window_city)) \
    .filter(col("rank") <= 3)

display(customer_city_ranked)
```

## Task 2: Running Total of Sales

```python

# Step 1: Join orders with payments
orders_payments = orders.join(payments, "order_id")

# Step 2: Extract date from timestamp
orders_payments = orders_payments.withColumn(
    "order_date",
    to_date("order_purchase_timestamp")
)

# Step 3: Daily sales
daily_sales = orders_payments.groupBy("order_date") \
    .agg(sum("payment_value").alias("daily_sales"))

# Step 4: Window for running total
window_date = Window.orderBy("order_date") \
    .rowsBetween(Window.unboundedPreceding, Window.currentRow)

# Step 5: Running total
daily_sales_running = daily_sales.withColumn(
    "running_total",
    sum("daily_sales").over(window_date)
)

display(daily_sales_running)
```

## Task 3: Top Products per Category

```python
# Step 1: Total sales per product
product_sales = order_items.groupBy("product_id") \
    .agg(sum("price").alias("total_sales"))

# Step 2: Join with products to get category
product_with_category = product_sales.join(products, "product_id")

# Step 3: Join with category translation
product_category = product_with_category.join(
    category_translation,
    "product_category_name"
).select(
    col("product_category_name_english").alias("category"),
    "product_id",
    "total_sales"
)

# Step 4: Window function
window_category = Window.partitionBy("category") \
    .orderBy(col("total_sales").desc())

# Step 5: Rank products
product_category_ranked = product_category.withColumn(
    "rank",
    dense_rank().over(window_category)
)

display(product_category_ranked)
```

## Task 4: Customer Lifetime Value

```python
# Step 1: Join payments with orders
orders_payments = orders.join(payments, "order_id")

# Step 2: Calculate total spend per customer (CLV)
customer_lifetime = orders_payments.groupBy("customer_id") \
    .agg(sum("payment_value").alias("customer_lifetime_value"))

display(customer_lifetime)
```

## Task 5: Customer Segmentation

```python
customer_segmented = customer_lifetime.withColumn(
    "segment",
    when(col("customer_lifetime_value") > 10000, "Gold")
    .when((col("customer_lifetime_value") >= 5000) & (col("customer_lifetime_value") <= 10000), "Silver")
    .otherwise("Bronze")
)

display(customer_segmented.select("customer_id", "customer_lifetime_value", "segment"))

segment_count = customer_segmented.groupBy("segment") \
    .agg(count("customer_id").alias("customer_count"))

display(segment_count)
```

## # Task 6: Final Reporting Table

```python
# Step 1: Orders per customer
customer_orders = orders.groupBy("customer_id") \
    .agg(count("order_id").alias("total_orders"))

# Step 2: Final report
final_report = customer_segmented \
    .join(customers, "customer_id") \
    .join(customer_orders, "customer_id") \
    .select(
        "customer_id",
        col("customer_city").alias("city"),
        col("customer_lifetime_value").alias("total_spend"),  # adjust if needed
        "segment",
        "total_orders"
    )

display(final_report)
```
